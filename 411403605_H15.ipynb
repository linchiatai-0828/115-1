```python
listA=[]
while True:
  item=input("輸入A倉文具清單，輸入 '/' 表示結束")
  if item=='/':
    break
  listA.append(item)
listB=[]
while True:
  item=input("輸入B倉文具清單，輸入 '/' 表示結束")
  if item=='/':
    break
  listB.append(item)
listC=listA+listB
listC.append("膠水")
listC="/".join(listC)
print("二倉合併清單：",listC)
```
